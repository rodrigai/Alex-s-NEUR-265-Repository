```python
#Imported NumPy, matplotlib, and statistics for usage in later steps

import numpy as np

from matplotlib import pyplot as plt

import statistics as st

%whos
```

```python
# Made variables into numphy arrays by reading data files

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
v_rest1 = np.loadtxt(url)
print(type(v_rest1))
print(v_rest1)

url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
v_rest2 = np.loadtxt(url2)
print(type(v_rest2))
print(v_rest2)
```

```python
#Concatenates v_rest_sst and v_rest_pvalb into a matrix with 2 rows and 500 columns named v_rest

v_rest_sst = v_rest1
v_rest_pvalb = v_rest2

v_rest = np.array([v_rest_sst, v_rest_pvalb])
print(v_rest)

np.shape(v_rest)
```

```python
#Prints size of v_rest

v_rest.size
```

```python
# Prints first row of v_rest

print(v_rest[0])
```

```python
# Returns sum of values in the 3rd column of v_rest

v_rest[:, 2].sum()
```

```python
#Returns 2nd to last value in the 2nd row of v_rest (confused as to why not recieving value)

v_rest[1:-2]
```

```python
# Makes a histogram with values of first row in v_rest with axes labeled

row1 = v_rest[0]

plt.hist(row1)

plt.xlabel('Membrane Potential')
plt.ylabel('Count')
```

```python
# Takes mean of values in first row of v_rest and assigns to v_rest_mean_sst
# Takes mean of values in second row of v_rest and assigns to v_rest_mean_pvalb

v_rest_mean_sst = st.mean(row1)
print(v_rest_mean_sst)

row2 = v_rest[1]
v_rest_mean_pvalb = st.mean(row2)
print(v_rest_mean_pvalb)
```

Between the two rows, the mean of the first row is greater than the mean of the second row by about 2. While they are close in value, they are negative values, so the lower mean is therefore greater.

```python
#Imports scipy

import scipy.stats as stats

%who
```

```python
#Performs a t-test on the two rows of v_rest

test = stats.ttest_ind(row1, row2)
print(test)
```

Results of my t-test displayed that the means are likely not significantly different, as my p = 7.16. This is greater than 5%.
