```python
#Imports relevant modules

import numpy as np
import matplotlib.pyplot as plt

%whos
```

```python
#Imports appropriate files as numpy arrays

url ='https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv'
acc_hits = np.loadtxt(url, delimiter = ',')
print(acc_hits)

url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv'
acc_fa = np.loadtxt(url2, delimiter = ',')
print(acc_fa)
```

```python
#Prints shape of each array

print(acc_hits.shape)
print(acc_fa.shape)
```

```python
#Create time variable with same length a acc_hits and acc_fa ith equally spaced values between 0 and 4

time = np.linspace(0, 4, 8001)
print(time)
```

```python
#Creates an expression to determine the sampling rate of the data

sampling_rate = (8001/4)
print(sampling_rate)
```

```python
#Plots first trial of acc_hits and acc_fa in different subplots, puts time on the x-axis, and labels graph axes

fig, ax = plt.subplots(2,1)
ax[0].plot(time, acc_hits[0])
ax[1].plot(time, acc_fa[0])

ax[0].set_xlabel('Time [s]')
ax[0].set_ylabel('Voltage []')

ax[1].set_xlabel('Time [s]')
ax[1].set_ylabel('Voltage []')
```

I think the dominant ocillation is 1.25 Hz for this LFP, as there were 5 peaks total over the course of 4 seconds (5/4 = 1.25). This did not differ between the first trial of acc_hits and that of acc_fa, as they both had 5 total peaks over the course of the 4 length signal.

```python
#Calculates and plots the ERP for acc_hits with time on the x-axis. Puts a black line at the "two" position on the x-axis and at the "0" position on the y-axis

plt.plot(time, acc_hits.mean(0))        # Plot the ERP of acc_hits
plt.xlabel('Time [s]')           # Label the axes
plt.ylabel('Voltage []')
plt.title('ERP of acc_hits')  # ... provide a title

plt.vlines(2, 54, -50, color = 'black') #Know we talked about keeping the limits the same, but looked different on singular plot rather than subplot
plt.hlines(0, 0, 4, color = 'black')
```

No, I do not see consistent activity in the LFP surrounding screen touch. Each side of the graph, marked by the vertical line, displays a different voltage pattern. The left side of the graph has a lower voltage with few peaks above or below the horizontal line. The right side of the graph, on the other hand, has also few peaks but they have a higher voltage and therefore span a greater distance above and below the horizontal black line.

```python
#Calculates and plots the ERP for acc_fa with time on the x-axis. Puts a black line at the "two" position on the x-axis and at the "0" position on the y-axis

plt.plot(time, acc_fa.mean(0))        # Plot the ERP of acc_fa
plt.xlabel('Time [s]')           # Label the axes
plt.ylabel('Voltage []')
plt.title('ERP of acc_fa')  # ... provide a title

plt.vlines(2, 54, -50, color = 'black')
plt.hlines(0, 0, 4, color = 'black')
```

Unlike the acc_hits, the ERP of acc_fa does show consistent activity in the LFP surrounding screen touch. Both sides divided by the vertical black line demonstrate a consistent amount of activity, which is very different than the large peaks on the right side of the acc_hits plot. It can therefore be assumed that there is more consistent activity in the LFP during false alarms than hits.
