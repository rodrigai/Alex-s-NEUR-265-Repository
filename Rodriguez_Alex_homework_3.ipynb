```python
#Imports numPy, matplotlib, and statistics for usage in later steps

import numpy as np

import matplotlib.pyplot as plt

import scipy.stats as stats

%whos
```

```python
# Makes data variables into numphy arrays

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
v_rest_sst = np.loadtxt(url)
print(type(v_rest_sst))
print(v_rest_sst)

url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
v_rest_pvalb = np.loadtxt(url2)
print(type(v_rest_pvalb))
print(v_rest_pvalb)
```

```python
# Finds means of v_rest_sst and v_rest_pvalb

mean_v_rest_sst = np.mean(v_rest_sst)
print(mean_v_rest_sst)

mean_v_rest_pvalb = np.mean(v_rest_pvalb)
print(mean_v_rest_pvalb)

# Performs a t-test to test the null hypothesis that the means of v_rest_sst and v_rest_pvalb come from different distributions
test = stats.ttest_ind(v_rest_sst, v_rest_pvalb)
print(test)

# Plots seperate histograms of v_rest_sst and v_rest_pvalb if the p-value of the t-test is < 0.05

if test.pvalue < 0.05:
  fig, ax = plt.subplots(2,1)
  ax[0].hist(v_rest_sst, color = 'paleturquoise')
  ax[1].hist(v_rest_pvalb, color ='lavender')

# Labels the x and y axes of the histograms

ax[0].set_xlabel('Resting Potential')
ax[0].set_ylabel('Frequency')

ax[1].set_xlabel('Resting Potential')
ax[1].set_ylabel('Frequency')
```
