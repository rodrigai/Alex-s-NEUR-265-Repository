```python
#Imported matplotlib, Pandas, and NumPy for usage in later steps

from matplotlib import pyplot as plt

import pandas as pd

import numpy as np

%whos
```

```python
#Imported v_rest_sst.csv file and turned it into Pandas DataFrame

URL= 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
pandas_data1= pd.read_csv(URL)
print(pandas_data1)
```

```python
#Imported v_rest_pvalb.csv file and turned it into Pandas DataFrame

URL2= 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
pandas_data2 = pd.read_csv(URL2)
print(pandas_data2)
```

```python
#Converted Pandas DataFrame to a list variable

data1= pandas_data1.values.tolist()
print(data1)

data2= pandas_data2.values.tolist()
print(data2)
```

```python
#Put brackets around all elements of Pandas DataFrame instead of each element individually

data1= np.squeeze(data1)
print(data1)
```

```python
#Puts brackets around all elements of Pandas DataFrame instead of each element individually

data2= np.squeeze(data2)
print(data2)
```

```python
#Printed entire length of data1

len(data1)
```

```python
#Prints entire length of data2 to help me create interval I want for slices

len(data2)
```

```python
#Prints 3rd, 4th, 5th, and 6th elements of data2

print(data2[-497:-493])
```

```python
#Prints the second-to-last and last elements of data1

print(data1[-2:])
```

```python
#Checks type of data2 vector

print(data2.dtype)
```

```python
#Makes histogram of data1

plt.hist(data1)
```

```python
#Re-makes histogram and labels function axes

plt.hist(data1)

plt.xlabel('Resting Potential')
plt.ylabel('Frequency')
```

```python
#Makes histograms and labels function axes for data2

plt.hist(data2)

plt.xlabel('Resting Potential')
plt.ylabel('Frequency')
```

data1's histogram appears to have a lower resting membrane potential of about -38 which data2's histogram appears to have a higher resting membrane potential of about -20. Additionally, data2's histogram has a higher frequency range with a max of 140, while data1's histogram has a lower max frequency range of 120.
